# Tableau CSV Exporter

Exports the gold-layer tables as CSV files to a Unity Catalog volume for use in Tableau Public. Since `tableauhyperapi` cannot install on ARM64 serverless compute, CSV is the simplest path to get data into Tableau.

```python
# Create volume for Tableau CSV exports
spark.sql("CREATE VOLUME IF NOT EXISTS workspace.nj_transit_gold.tableau_exports")

VOLUME_PATH = "/Volumes/workspace/nj_transit_gold/tableau_exports"
GOLD_SCHEMA = "workspace.nj_transit_gold"

print(f"Volume ready: {VOLUME_PATH}")
```

```python
# Export enriched_trips as a single CSV file
# coalesce(1) ensures one file for easy download into Tableau Public
spark.table(f"{GOLD_SCHEMA}.enriched_trips") \
    .coalesce(1) \
    .write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv(f"{VOLUME_PATH}/enriched_trips")

print("enriched_trips exported")
```

```python
# Export station_weather_region as a single CSV file
spark.table(f"{GOLD_SCHEMA}.station_weather_region") \
    .coalesce(1) \
    .write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv(f"{VOLUME_PATH}/stations")

print("stations exported")
```

```python
# Export line_coordinates as a single CSV file
spark.table(f"{GOLD_SCHEMA}.line_coordinates") \
    .coalesce(1) \
    .write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv(f"{VOLUME_PATH}/line_coordinates")

print("line_coordinates exported")
```
